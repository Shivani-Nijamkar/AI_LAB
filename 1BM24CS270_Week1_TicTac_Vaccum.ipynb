```python
# Tic-Tac-Toe Game
# Human Player = X
# Computer Player = O

board = [' '] * 9


# Display the board
def display_board():
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()


# Check whether a player has won
def check_winner(player):
    winning_positions = [
        (0, 1, 2),  # Row 1
        (3, 4, 5),  # Row 2
        (6, 7, 8),  # Row 3
        (0, 3, 6),  # Column 1
        (1, 4, 7),  # Column 2
        (2, 5, 8),  # Column 3
        (0, 4, 8),  # Diagonal
        (2, 4, 6)   # Diagonal
    ]

    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


# Check whether the board is full
def board_full():
    return ' ' not in board


# Computer's move
def computer_move():
    # First, try to win
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'

            if check_winner('O'):
                return

            board[i] = ' '

    # Otherwise, block the human player
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'

            if check_winner('X'):
                board[i] = 'O'
                return

            board[i] = ' '

    # Otherwise, choose the first available position
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            return


# Main game
print("TIC-TAC-TOE")
print("Human Player = X")
print("Computer Player = O")

while True:

    # Display the board
    display_board()

    # Human player's move
    position = int(input("Enter position (1-9): ")) - 1

    # Validate the move
    if position < 0 or position > 8 or board[position] != ' ':
        print("Invalid position. Try again.")
        continue

    board[position] = 'X'

    # Check if human wins
    if check_winner('X'):
        display_board()
        print("Human Player Wins!")
        break

    # Check for draw
    if board_full():
        display_board()
        print("Game Draw!")
        break

    # Computer's move
    computer_move()
    print("Computer played.")

    # Check if computer wins
    if check_winner('O'):
        display_board()
        print("Computer Player Wins!")
        break

    # Check for draw
    if board_full():
        display_board()
        print("Game Draw!")
        break
```

```python
import random

# Randomly generate the initial state
rooms = {
    'A': random.choice(['Clean', 'Dirty']),
    'B': random.choice(['Clean', 'Dirty'])
}

# Random starting position
position = random.choice(['A', 'B'])

print("Initial State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Position:", position)

# Vacuum cleaner agent
while rooms['A'] == 'Dirty' or rooms['B'] == 'Dirty':

    # If current room is dirty, clean it
    if rooms[position] == 'Dirty':
        print("\nAction: SUCK")
        rooms[position] = 'Clean'
        print("Room", position, "is now Clean")

    # If current room is clean, move to the other room
    else:
        if position == 'A':
            position = 'B'
        else:
            position = 'A'

        print("\nAction: MOVE")
        print("Vacuum moved to Room", position)

# Final state
print("\nFinal State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Position:", position)
print("Goal Achieved: Both rooms are Clean")
```
