```python
import itertools

def evaluate_implies(p, q):
    """Evaluates logical implication: P => Q is equivalent to (not P or Q)"""
    return (not p) or q

def tt_entails(symbols, kb_func, query_func):
    """
    Truth-Table Enumeration Algorithm for Propositional Entailment.

    Parameters:
    - symbols: List of propositional symbol names (e.g., ['A', 'B'])
    - kb_func: Lambda function representing the Knowledge Base KB(model)
    - query_func: Lambda function representing the Query alpha(model)

    Returns:
    - True if KB entails alpha, False otherwise.
    """
    n = len(symbols)
    all_models = list(itertools.product([True, False], repeat=n))

    # Header for Truth Table
    header = " | ".join(symbols) + " | KB | Query | KB => Query"
    print(header)
    print("-" * len(header))

    entails = True

    for model_values in all_models:
        # Create dictionary mapping symbol name to True/False value
        model = dict(zip(symbols, model_values))

        kb_val = kb_func(model)
        query_val = query_func(model)

        # Check entailment condition: whenever KB is True, Query must be True
        valid_row = True
        if kb_val and not query_val:
            entails = False
            valid_row = False

        # Format values for neat printing
        row_str = " | ".join(["T" if model[s] else "F" for s in symbols])
        kb_str = "T" if kb_val else "F"
        query_str = "T" if query_val else "F"
        status_str = "Valid" if (kb_val and query_val) else ("Invalid" if not valid_row else "—")

        print(f"{row_str} |  {kb_str} |   {query_str}   | {status_str}")

    return entails


# Driver Code
if __name__ == "__main__":
    # Example from image:
    # KB = (A => B) and A
    # Query alpha = B

    symbols = ['A', 'B']

    # Define KB using lambda taking dictionary 'm' representing truth assignment model
    kb = lambda m: evaluate_implies(m['A'], m['B']) and m['A']

    # Define Query
    query = lambda m: m['B']

    print("--- Truth Table Enumeration ---")
    result = tt_entails(symbols, kb, query)

    print("-" * 35)
    if result:
        print("Conclusion: KB |= Query (KB entails Query)")
    else:
        print("Conclusion: KB does NOT entail Query")
```
