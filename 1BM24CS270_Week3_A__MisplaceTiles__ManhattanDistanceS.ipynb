```python
#week3(misplaced tiles)
import heapq

# Initial and Goal states specified in the lab assignment
INITIAL_STATE = (
    (2, 8, 3),
    (1, 6, 4),
    (0, 7, 5)  # 0 represents the blank space
)

GOAL_STATE = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

def find_blank(state):
    """Finds row and column of blank space (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generates valid next states by sliding adjacent tiles into the blank space."""
    r, c = find_blank(state)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right
    neighbors = []

    for dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors

def h_misplaced(state, goal=GOAL_STATE):
    """Heuristic: Count of misplaced non-blank tiles."""
    count = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0 and val != goal[r][c]:
                count += 1
    return count

def a_star_misplaced(initial_state, goal_state):
    """A* Search Algorithm using Misplaced Tiles Heuristic."""
    open_list = []
    # Priority Queue elements: (f, g, current_state, path)
    heapq.heappush(open_list, (h_misplaced(initial_state), 0, initial_state, [initial_state]))
    visited = set()

    while open_list:
        f, g, current, path = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)

        if current == goal_state:
            return path, len(visited)

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                g_new = g + 1
                h_new = h_misplaced(neighbor, goal_state)
                f_new = g_new + h_new
                heapq.heappush(open_list, (f_new, g_new, neighbor, path + [neighbor]))

    return None, len(visited)

def print_state(state):
    """Prints single board state cleanly."""
    for row in state:
        print(" ".join(str(cell) if cell != 0 else "_" for cell in row))
    print()

if __name__ == "__main__":
    path, nodes_visited = a_star_misplaced(INITIAL_STATE, GOAL_STATE)

    print("================ CASE 1: Misplaced Tiles Heuristic ================")
    print(f"Total Steps (Cost): {len(path) - 1}")
    print(f"Total Nodes Visited: {nodes_visited}\n")

    for step, state in enumerate(path):
        print(f"Step {step}:")
        print_state(state)
```

```python
#week 3 -part2 (manhatten distance)
import heapq

# Initial and Goal states specified in the lab assignment
INITIAL_STATE = (
    (2, 8, 3),
    (1, 6, 4),
    (0, 7, 5)  # 0 represents the blank space
)

GOAL_STATE = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

def find_blank(state):
    """Finds row and column of blank space (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generates valid next states by sliding adjacent tiles into the blank space."""
    r, c = find_blank(state)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # Up, Down, Left, Right
    neighbors = []

    for dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors

def h_manhattan(state, goal=GOAL_STATE):
    """Heuristic: Sum of Manhattan distances for non-blank tiles."""
    goal_pos = {}
    for r in range(3):
        for c in range(3):
            goal_pos[goal[r][c]] = (r, c)

    distance = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0:
                gr, gc = goal_pos[val]
                distance += abs(r - gr) + abs(c - gc)
    return distance

def a_star_manhattan(initial_state, goal_state):
    """A* Search Algorithm using Manhattan Distance Heuristic."""
    open_list = []
    # Priority Queue elements: (f, g, current_state, path)
    heapq.heappush(open_list, (h_manhattan(initial_state), 0, initial_state, [initial_state]))
    visited = set()

    while open_list:
        f, g, current, path = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)

        if current == goal_state:
            return path, len(visited)

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                g_new = g + 1
                h_new = h_manhattan(neighbor, goal_state)
                f_new = g_new + h_new
                heapq.heappush(open_list, (f_new, g_new, neighbor, path + [neighbor]))

    return None, len(visited)

def print_state(state):
    """Prints single board state cleanly."""
    for row in state:
        print(" ".join(str(cell) if cell != 0 else "_" for cell in row))
    print()

if __name__ == "__main__":
    path, nodes_visited = a_star_manhattan(INITIAL_STATE, GOAL_STATE)

    print("================ CASE 2: Manhattan Distance Heuristic ================")
    print(f"Total Steps (Cost): {len(path) - 1}")
    print(f"Total Nodes Visited: {nodes_visited}\n")

    for step, state in enumerate(path):
        print(f"Step {step}:")
        print_state(state)
```
