```python
import random

def calculate_heuristic(state):
    """
    Calculates the total number of attacking pairs of queens (row and diagonal conflicts).
    State representation: state[i] is the row index of the queen in column i.
    """
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check row conflict
            if state[i] == state[j]:
                attacks += 1
            # Check diagonal conflict
            elif abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_neighbors(state):
    """
    Generates all neighbor states by moving one queen at a time
    to every other row in its column.
    """
    n = len(state)
    neighbors = []

    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(neighbor)

    return neighbors

def hill_climbing(initial_state):
    """
    Executes the Hill Climbing algorithm.
    """
    current_state = list(initial_state)
    n = len(current_state)
    step = 0

    print(f"Initial State: {current_state} | Initial Cost h(S): {calculate_heuristic(current_state)}\n")

    while True:
        current_cost = calculate_heuristic(current_state)

        # Step 3: Check Goal
        if current_cost == 0:
            print(f"Goal State Reached: {current_state} with cost h(S) = 0!")
            return current_state

        # Step 4: Generate Neighbors
        neighbors = get_neighbors(current_state)

        # Step 5: Find Best Neighbor
        best_neighbor = None
        best_cost = current_cost

        for neighbor in neighbors:
            cost = calculate_heuristic(neighbor)
            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor

        # Step 6: Decision & Transition
        if best_cost < current_cost:
            step += 1
            current_state = best_neighbor
            print(f"Step {step}: Moved to State {current_state} | Cost h(S) = {best_cost}")
        else:
            print(f"\nStopped! Stuck at Local Minimum: {current_state} | Cost h(S) = {current_cost}")
            return current_state

def print_board(state):
    """Prints a visual N x N grid representation of the board."""
    n = len(state)
    print("\nBoard Visualization:")
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

# Driver Code
if __name__ == "__main__":
    # Initial configuration for 4-Queens
    initial_board = [3, 1, 2, 0]

    final_state = hill_climbing(initial_board)
    print_board(final_state)
```
